**Ejercicio 1**

Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

```python
from itertools import product

S = list(product(range(1, 7), repeat=2)) #las 36 combinaciones que se dan al multiplicar los 2 dados
print("Espacio Muestral:", S)

A = [x for x in S if x[0] + x[1] == 5] #combinaciones que dan 5
print("Casos favorables:", A)

print("Probabilidad:", round (len(A) / len(S), 2))
```

**Ejercicio 2**

En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.



```python
from itertools import product

S = list(product(range(10), repeat=3)) #contraseñas formadas con 3 dijitos del 0 al 9 que se pueden hacer
print("Espacio Muestral:", S)

A = [x for x in S if len(set(x)) == 3]
print("Casos favorables:", len(A)) #ningun dijito se repite

print("Casos posibles:", len(S))
print("Probabilidad:", len(A)/len(S))
```

**Ejercicio 3**

Tienes 8 canciones distintas y las pones en un orden aleatorio, con todos los órdenes igualmente probables. Calcula la probabilidad de que tu favorita quede primera y la que menos te gusta quede última.

```python
from itertools import permutations

canciones = [1, 2, 3, 4, 5, 6, 7, 8]
S = list(permutations(canciones))
A = [x for x in S if x[0] == 1 and x[7] == 8]

print("Resultados posibles:", len(S))
print("Casos favorables:", len(A))
print("Probabilidad:", len(A)/len(S))
print("Probabilidad:", round(len(A) / len(S) * 100, 2), "%")
```

**Ejercicio 4**

Se lanzan tres dados equilibrados. ¿Qué es más probable, que la suma sea 9 o que la suma sea 10? ¿Por cuántos resultados le gana una suma a la otra?

```python
from itertools import product

S = list(product(range(1, 7), repeat=3))
print("Espacio Muestral:", len(S))

A = [x for x in S if sum(x) == 9]
B = [x for x in S if sum(x) == 10]

print("---------probabilidad de 9----------")
print("casos favorables de la suma de 9:", len(A))
print("Probabilidad suma 9:", len(A)/len(S))
print("Probabilidad suma 9:", round(len(A)/len(S)*100, 2), "%")

print("---------probabilidad de 10----------")
print("Casos favorables suma 10:", len(B))
print("Probabilidad suma 10:", len(B)/len(S))
print("Probabilidad suma 10:", round(len(B)/len(S)*100, 2), "%")


print("¿Por cuántos resultados le gana una suma a la otra? por:", len(B) - len(A))
print("la suma mas problable es:", 10)
```

**Ejercicio 5**

Ana, Bruno, Carla y Diego hacen un amigo secreto: cada quien saca un regalo al azar y todos los repartos son igualmente probables. Calcula la probabilidad de que nadie saque su propio regalo.

```python
from itertools import permutations

amigos = ["Ana", "Bruno", "Carla", "Diego"]
S = list(permutations(amigos))

A = [x for x in S if all(x[i] != amigos[i] for i in range(4))]

print("Resultados posibles:", len(S))
print("Casos favorables:", len(A))
print("Probabilidad de no sacar su propio regalo:", round(len(A) / len(S) * 100, 2), "%")
```
