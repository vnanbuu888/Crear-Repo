**Ejercicio 1**

Simula 10 000 lanzamientos de dos dados y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.

```python
import random

#Funcion que simula el lanzamiento de dos dados
def lanzar_dados():
    dado1 = random.randint(1, 6)
    dado2 = random.randint(1, 6)
    return dado1 + dado2

n = 10000
conteo = 0 #las veces que la suma es 5

for i in range(n):
    resultado = lanzar_dados()

    if resultado == 5:
        conteo += 1


proba_frecuentista = conteo / n
proba_clasica = 4 / 36

print("Numero de lanzamientos:", n)
print("Casos favorables:", conteo)
print("Probabilidad frecuentista:", round(proba_frecuentista * 100, 2), "%")
print("Probabilidad clasica:", round(proba_clasica * 100, 2), "%")
```

**Ejercicio 2**

Simula 10 000 PIN de 4 dígitos, del 0 al 9, y estima la probabilidad de que ninguno se repita. Compara el resultado con la fracción de la parte clásica.

```python
import random

def generar_pin():
    return [random.randint(0, 9) for _ in range(4)]

n = 10000
conteo = 0

for i in range(n):
    pin = generar_pin()

    if len(set(pin)) == 4:
        conteo += 1

proba_frecuentista= conteo / n
proba_clasica = (10 * 9 * 8 * 7) / (10 ** 4)

print("Numero de experimentos:", n)
print("Casos favorables:", conteo)
print("Probabilidad frecuentista:", round(proba_frecuentista * 100, 2), "%")
print("Probabilidad clasica:", round(proba_clasica * 100, 2), "%")
```

**Ejercicio 3**

Simula 20 000 lanzamientos de tres dados. Estima la frecuencia de la suma 9 y la de la suma 10. ¿Se nota, en la simulación, cuál de las dos sale más?

```python
import random

def lanzar_dado():
    return random.randint(1, 6)

n = 20000
conteo_9 = 0
conteo_10 = 0

for i in range(n):
    dado1 = lanzar_dado()
    dado2 = lanzar_dado()
    dado3 = lanzar_dado()

    suma = dado1 + dado2 + dado3

    if suma == 9:
        conteo_9 += 1
    elif suma == 10:
        conteo_10 += 1

frecuencia_9 = conteo_9 / n
frecuencia_10 = conteo_10 / n

print("Numero de lanzamientos:", n)
print("Casos con suma 9:", conteo_9)
print("Frecuencia de la suma 9:", round(frecuencia_9 * 100, 2), "%")

print("Casos con suma 10:", conteo_10)
print("Frecuencia de la suma 10:",round(frecuencia_10 * 100, 2), "%")


if conteo_10 > conteo_9:
    print("La suma 10 sale mas veces.")
elif conteo_9 > conteo_10:
    print("La suma 9 sale mas veces.")
else:
    print("Ambas sumas salieron la misma cantidad.")
```

**Ejercicio 5**

Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

```python
import random

def lanzar_moneda():
    return random.choice(["águila", "sol"])

n = 10000
s_favorables = 0

for i in range(n):
    racha = 0
    encontro_5 = False

    for j in range(20):
        resultado = lanzar_moneda()

        if resultado == "águila":
            racha += 1
        else:
            racha = 0

        if racha >= 5:
            encontro_5 = True

    if encontro_5:
        s_favorables += 1

pro_frecuencia = s_favorables / n

print("Número de series:", n)
print("Probabilidad:", round(pro_frecuencia * 100, 2), "%")
```
