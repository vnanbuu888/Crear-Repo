**Ejercicio A-1**

Se lanza un dado equilibrado. Calcula la probabilidad de obtener un 2, dado que salió un número par.

```python
from itertools import product

B = list(product(range(1, 7), repeat=1))
A = [x for x in B if x[0] % 2 == 0]
# numeros pares A={2,4,6}

print("Resultados posibles:", len(B))
print("Casos favorables:", A)
print("Probabilidad:", 1 / len(A))
```

**Ejercicio A-2**

Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

```python
from itertools import product

B = list(product(range(1,7), repeat=3))
B = [x for x in B if len(set(x)) == 3]
A = [x for x in B if sum(x) == 9]

print("Resultados posibles:", len(B))
print("Casos favorables distintos:", len(A))
print("Probabilidad:", len(A) / len(B))
```

De 100 días: 40 son nublados De esos, 30 llueve
¿Cuál es la probabilidad de lluvia dado que está nublado?

**Ejercicio C-1**

El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta», ¿cuál es la probabilidad de que sea spam?

```python
S = 0.20 * 0.80
D = 0.80 * 0.10

P = S / (S + D)

print("correos con spam:", round(S, 2))
print("correo sin spam:", round(D, 2))
print("Probabilidad de que si es spam:", round(P, 2))
```

**Ejercicio C-2**

En una tienda online, el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.

Si se registró una venta, ¿qué navegador es más probable que haya usado el cliente?

```python
C = 0.40 * 0.03
S = 0.35 * 0.04
O = 0.25 * 0.02

P = C + S + O

P_C = C / P
P_S = S / P
P_Otros = O / P

print("Chrome:", round(P_C, 2))
print("Safari:", round(P_S, 2))
print("otros navegadores:", round(P_Otros, 2))

print("Navegador más probable que uso el cliente: Safari")
```
